```python

import numpy as np
```

```python
a = np.array([1, 2, 3, 4])
np.all(a != 0)
# ninguno de los elementos es cero
```

```python
a = np.array([0, 0, 2, 0])
np.any(a != 0)
# existe al menos un elemento distinto de cero
```

```python
a = np.array([1, 2, np.nan, np.inf])
np.isnan(a), np.isinf(a)
# detecta valores NaN e infinitos
```

```python
a = np.array([1, 2, np.inf])
np.isinf(a).any()
# contiene valores infinitos
```

```python
a = np.array([1, np.nan, 3])
np.isnan(a).any()
# contiene valores NaN
```

```python
a = np.array([1, 3, 5])
b = np.array([2, 3, 4])
a > b, a >= b, a < b, a <= b
# comparación elemento a elemento
```

```python
a = np.array([1, 7, 13, 105])
a.nbytes
# memoria ocupada en bytes
```
